# Smart Attendance Logger

```python
def mark_attendance():
    final_names=set()
    while True:
            
        name=input("Please enter student name: ")
        if name=="STOP":
            break
        else:
            final_names.add(name)

    print(final_names)
    print (len(final_names) ) 

```

```python
mark_attendance()
```

# E-Commerce Discount Engine

```python

total_purchase_amt=float(input("Enter Total Purchase amount: "))
def discount_calc(total_purchase_amt):
    discount1=total_purchase_amt*0.2
    discount2=total_purchase_amt*0.1
    if total_purchase_amt > 500:
        return(f"Discount Price is: {discount1} ")
    elif total_purchase_amt > 200:
        return(f"Discount Price is: {discount2}")
    else:
        return("No Discount for total purchase below 200...")


discount_calc(total_purchase_amt)   
```
